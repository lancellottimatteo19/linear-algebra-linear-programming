# Linear Algebra and Linear Programming

Core numerical methods for optimisation: singular value decomposition and least squares, and constrained linear programming solved and visualised geometrically.

---
## 1. Setup

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import linprog

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 2. SVD and least squares

```python
np.random.seed(42)
X_data = np.linspace(0, 10, 50)
y_data = 2.5 * X_data + 1.0 + np.random.normal(0, 2.0, size=X_data.shape)

A = np.column_stack([np.ones_like(X_data), X_data])
b = y_data

# Singular value decomposition
U, S, Vt = np.linalg.svd(A, full_matrices=False)
print("Singular values (S):", S)
print(f"Condition number: {S[0] / S[-1]:.2f}")
print("Reconstruction A = U diag(S) V^T is exact:", np.allclose(A, U @ np.diag(S) @ Vt))

# Least squares via the pseudo-inverse: x = V diag(1/S) U^T b
x_svd = Vt.T @ ((U.T @ b) / S)

# Least squares with numpy, for comparison
x_lstsq, _, _, _ = np.linalg.lstsq(A, b, rcond=None)
print(f"Intercept: {x_lstsq[0]:.4f} | Slope: {x_lstsq[1]:.4f}")
print("SVD solution equals lstsq:", np.allclose(x_svd, x_lstsq))

plt.figure()
plt.scatter(X_data, y_data, color='crimson', label='Observed data')
plt.plot(X_data, A @ x_lstsq, color='navy', linewidth=2.5, label='Least-squares fit')
plt.title('Fitting noisy data with least squares')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## 3. Linear programming in two dimensions

```python
# Maximise Z = 3*x0 + 5*x1  ==>  minimise -3*x0 - 5*x1
c = [-3, -5]
A_ub = [[1, 2], [4, 2]]
b_ub = [8, 16]

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=[(0, None), (0, None)], method='highs')
assert res.success, res.message
print(f"Optimum at (x0, x1) = ({res.x[0]:.2f}, {res.x[1]:.2f}) with value {-res.fun:.2f}")

# Feasible region
x0_vals = np.linspace(0, 5, 400)
y1 = (8 - x0_vals) / 2.0
y2 = (16 - 4*x0_vals) / 2.0

plt.figure()
plt.plot(x0_vals, y1, label=r'$x_0 + 2x_1 \leq 8$')
plt.plot(x0_vals, y2, label=r'$4x_0 + 2x_1 \leq 16$')
y_pass = np.minimum(y1, y2)
# Iso-profit line 3*x0 + 5*x1 = Z* through the optimum
plt.plot(x0_vals, (-res.fun - 3 * x0_vals) / 5, 'k:', label=f'Iso-profit line $Z = {-res.fun:.2f}$')
plt.fill_between(x0_vals, 0, np.maximum(0, y_pass), where=(y_pass >= 0), color='gray', alpha=0.3, label='Feasible region')
plt.scatter(res.x[0], res.x[1], color='red', s=120, zorder=5, label='Optimal solution')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.xlabel('$x_0$')
plt.ylabel('$x_1$')
plt.legend()
plt.title('Linear programming: 2D feasible region')
plt.show()
```
